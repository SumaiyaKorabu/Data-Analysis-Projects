## Q20. Display a dataframe of Purpose and the total distance travelled for that particular Purpose. (3 points)
#### Note: Use the original dataframe without dropping "NA" values

```python
# Group by 'Purpose' and sum the 'Distance' for each group
purpose_distance = uber_drives.groupby('PURPOSE*')['MILES*'].sum()

# Convert the Series to a DataFrame for better display
purpose_distance_df = purpose_distance.reset_index()

# Rename the columns for better understanding
purpose_distance_df.columns = ['Purpose', 'Total Distance (Miles)']

# Display the resulting DataFrame
purpose_distance_df
```

## Q21. Generate a plot showing count of trips vs category of trips. What can you infer from the plot (2 +1 points)
#### Note: Use the original dataframe without dropping the 'NA' values.

```python
# Count the number of trips for each category
trip_counts = uber_drives['CATEGORY*'].value_counts()

# Plotting the data using Seaborn for better aesthetics
plt.figure(figsize=(8, 4))
sns.barplot(x=trip_counts.index, y=trip_counts.values, palette='viridis')

# Add labels and title
plt.xlabel('Category of Trips')
plt.ylabel('Count of Trips')
plt.title('Count of Trips vs Category of Trips')

# Show the plot
plt.show()
```

Note: The count of trips is more for business category

## Q22. What percentage of Miles were clocked under Business Category and what percentage of Miles were clocked under Personal Category ? (3 points)

#### Note:Use the original dataframe without dropping the 'NA' values. 


```python
# Calculate total miles for each category (Business and Personal)
total_miles_business = uber_drives[uber_drives['CATEGORY*'] == 'Business']['MILES*'].sum()
total_miles_personal = uber_drives[uber_drives['CATEGORY*'] == 'Personal']['MILES*'].sum()

# Calculate the overall total miles (including NA values)
total_miles = uber_drives['MILES*'].sum()

# Calculate the percentage of miles clocked under each category
percentage_miles_business = (total_miles_business / total_miles) * 100
percentage_miles_personal = (total_miles_personal / total_miles) * 100

# Display the results
print(f"Percentage of miles clocked under Business category: {percentage_miles_business:.2f}%")
print(f"Percentage of miles clocked under Personal category: {percentage_miles_personal:.2f}%")
```
