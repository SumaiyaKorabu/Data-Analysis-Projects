## Q 5. Find the net revenue generated by the E-Commerce company over all orders placed. (6 marks)

### Ans 5

```python
# Calculate number of orders per Brand
num_orders = ECom_Data.groupby('Brand')['Customer_uniq_id'].count().reset_index()
num_orders.columns = ['Brand', 'num_orders']
num_orders
```

```python
#Merging the number of orders by brand DataFrame with the original DataFrame
ECom_Data = pd.merge(ECom_Data, num_orders, on='Brand', how='left')
```

```python
ECom_Data.head()
```

```python
#Calculating the revenue for each order based on the rules

def cal_revenue(discounted_price, num_orders):
    if num_orders > 600:
        return 0.25 * discounted_price
    elif num_orders > 350:
        return 0.15 * discounted_price
    elif num_orders > 100:
        return 0.10 * discounted_price
    else:
        return 0.05 * discounted_price
```

```python
#Calculate revenue for each order

ECom_Data['orders_discounted_price'] = ECom_Data.apply(lambda row: cal_revenue(row['discounted_price'], row['num_orders']), axis=1)
```

```python
ECom_Data["Brand_revenue"]=ECom_Data["retail_price"]-ECom_Data['orders_discounted_price']
```

```python
ECom_Data.head()
```

```python

#total net revenue generated by the E-Commerce company over all orders placed

total_revenue = ECom_Data['Brand_revenue'].sum()
print("total net revenue generated by the E-Commerce company over all orders placed:", total_revenue)
```

## Q 9.Compare performance regionwise

## a)	Draw a lineplot for the monthly Revenue of ECom Company for each region separately.  (4 marks) 

## b)	Identify the best and the worst performing months for each region. (2 marks)


### Ans 9 a)

```python
#Converting date columns in dataset to datetime format
date_columns = ['Order_Date','Expected_Delivery_Date','Delivered_Date']
```

```python
for col in date_columns:
    ECom_Data[col]=pd.to_datetime(ECom_Data[col],dayfirst=True)
```

```python
ECom_Data.info()
```

```python
#Extracting month from Order_Date
ECom_Data['Order_Month'] = ECom_Data['Order_Date'].dt.to_period('M').astype(str)
```

```python
ECom_Data.head()
```

```python
#Calculating revenue for each region on monthly base
discounted_price = ECom_Data.groupby(['Region', 'Order_Month'])['discounted_price'].sum().reset_index(name="total_discount")
retail_price = ECom_Data.groupby(['Region', 'Order_Month'])['retail_price'].sum().reset_index(name="total_retail")
Monthly_revenue= retail_price["total_retail"]-discounted_price["total_discount"]
```

```python
Monthly_Regional_revenue = pd.DataFrame({'Month':retail_price['Order_Month'],'Region': retail_price['Region'], 'Monthly_revenue': Monthly_revenue})
Monthly_Regional_revenue
```

```python
# Plotting lineplot for the monthly Revenue of ECom Company for each region separately

plt.figure(figsize=(12, 6))
sns.lineplot(data=Monthly_Regional_revenue, x='Month', y='Monthly_revenue', hue='Region')
plt.title('Monthly Revenue of E-Commerce Company by Region')
plt.xlabel('Month')
plt.ylabel('Revenue')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

### Ans 9 b)

```python
#Finding best months

best_months = Monthly_Regional_revenue.groupby('Region').apply(lambda x: x.loc[x['Monthly_revenue'].idxmax()]['Month']).reset_index(name='Best_Month')
best_months
```

```python
#Finding worst months

worst_months = Monthly_Regional_revenue.groupby('Region').apply(lambda x: x.loc[x['Monthly_revenue'].idxmin()]['Month']).reset_index(name='Worst_Month')
worst_months
```
